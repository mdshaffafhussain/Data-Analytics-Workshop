```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("Book1.xlsx")

df.dropna(subset=["CustomerID"], inplace=True)

df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

df["Year"] = df["InvoiceDate"].dt.year

df["TotalSales"] = df["Quantity"] * df["UnitPrice"]
```

```python
trend = df.groupby("InvoiceDate")[["Quantity","UnitPrice"]].sum()

trend.plot(figsize=(10,6))

plt.title("Quantity and Unit Price Trend")
plt.xlabel("Invoice Date")
plt.ylabel("Value")

plt.show()
```

```python
top_products = df.groupby("Description")["InvoiceNo"].count().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
top_products.plot(kind="barh")

plt.title("Top Products by Number of Invoices")
plt.xlabel("Invoice Count")
plt.ylabel("Product")

plt.show()
```

```python
year_invoice = df.groupby("Year")["InvoiceNo"].count()

plt.figure(figsize=(10,6))

plt.scatter(year_invoice.index, year_invoice.values)

plt.title("Invoice Count by Year")
plt.xlabel("Year")
plt.ylabel("Invoice Count")

plt.show()
```

```python
date_trend = df.groupby("InvoiceDate")[["InvoiceNo","Quantity"]].count()

date_trend.plot(kind="area", figsize=(10,6), alpha=0.6)

plt.title("Invoice and Quantity Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Count")

plt.show()
```

```python
year_sales = df.groupby("Year")["TotalSales"].sum()

plt.figure(figsize=(10,6))
year_sales.plot(kind="bar", color="orange")

plt.title("Yearly Sales Value")
plt.xlabel("Year")
plt.ylabel("Total Sales")

plt.show()
```

```python
year_data = df.groupby("Year").agg({
    "Quantity":"sum",
    "InvoiceNo":"count"
})

fig, ax1 = plt.subplots(figsize=(10,6))

ax1.bar(year_data.index, year_data["Quantity"], alpha=0.7)
ax1.set_ylabel("Quantity")

ax2 = ax1.twinx()
ax2.plot(year_data.index, year_data["InvoiceNo"], marker="o")

ax2.set_ylabel("Invoice Count")

plt.title("Quantity vs Invoice Count by Year")

plt.show()
```
